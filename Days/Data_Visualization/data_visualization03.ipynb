# Bar Chart
- Bar chart is not used for small amount of data.

```python
import matplotlib.pyplot as plt
```

```python
department = ["Computer Science,", "Electrical", "Civil", "Aerospace", "Computer Engineering", "AI/Ml", "Physics", "Mechanical"]
no_of_students = [120, 60, 90, 80, 100, 50, 20, 70]
```

```python
# plt.style.use('fivethirtyeight')
plt.bar(department, no_of_students, edgecolor = "black")

for i, value in enumerate(no_of_students):  # enumerate gives index with its values respectively
  plt.text(i, value + 1, value, ha = "center") # ha = horizontal alignment

plt.xlabel("Department")
plt.ylabel("Number of Students")
plt.title("Department vs Number of Students")

plt.tight_layout()
plt.show()
```

# Histogram


- In hist there is a single variable.
- The plot is made accordingly to the range of values and the frequency.

```python
plt.style.use('fivethirtyeight')
ages = [18, 19, 20, 21, 21, 22, 22, 23, 24, 24,
        25, 26, 26, 27, 28, 29, 30, 30, 31, 32]
plt.hist(ages, bins = 10, color = "lightgreen", edgecolor = "black")

plt.xlabel("Age Range")
plt.ylabel("Number of Students")
plt.title("Age distribution of students")

plt.tight_layout()
plt.show()
```
