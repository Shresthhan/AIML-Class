# Data Visualization
-> Visual Representation of Data

In this course we study 3 libraries for data visualization

## Matplotlib
-> One of the classic libraries for data visualization


1.   Most of it is based on ggplot.
2.   It is difficult to write but easy to understand NOTE: In the function pie() which is used to plot pie charts, one of the parameter is: autopct = "%1.1f%%".
1.   Difficult to remember as different functions have different parameters and different parameter names for same work.







## Seaborn


1.   It is the xtended version of Matplotlib.
2.   Easy to write and easy to understand.


1.   It makes plots with difficult syntax in matplotlib easy.
2.   Both matplolib and seaborn cannot be directly used in web programs.


1.   Both matplotlib and seaborn cannot create interactive visualizations easily.







## Plotly

1. Makes the creation of interactive visualization very easy -> Visualization with beautiful animations.
2. Can be directly used in websites.
3. Plotly JS (Firstly it was introduced in JS)
4. The library that we use in python is entirely based on Plotly JS and is simply called Plotly.
5. Gives maximum output in minimum lines of codes.
