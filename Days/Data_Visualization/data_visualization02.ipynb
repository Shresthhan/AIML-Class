# Line Plots vs Scatter Plots

## Line Plot

- We use line plots when any one of the two variable is continuous.
- Age, Growth of time in an experiment.

image.png

## Scatter Plot

- When both the variables are discrete than we use scatter plot.
- Used to look at the correlation and classification / Clustering. (KNN)

```python
import pandas as pd
df = pd.read_csv("2019-05-31-data.csv")
```

```python
df.sample(10)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.style.use("fivethirtyeight")
```

```python
view_count = df["view_count"]
likes = df["likes"]
ratio = df["ratio"]
```

```python
print(view_count)
print(likes)
print(ratio)
```

```python
plt.scatter(view_count, likes, c = ratio, cmap = "summer", edgecolor = "black", linewidth = 0.5, alpha = 0.7)

cbar = plt.colorbar()
cbar.set_label("Like/Dislike Ratio")

plt.xscale("log")
plt.yscale("log")
plt.title("Trending Youtube Videos")
plt.xlabel("view_count")
plt.ylabel("Total_likes")

plt.tight_layout()
plt.show()
```

- Dislike Increase -> Ratio decrease
- Like Increase -> Ratio Increase
